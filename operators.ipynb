```python
n=int(input())
mat=[list(map(int,input().split()))for i in range(n)]
rot=[[0]*n for i in range(n)]
for i in range(n):
  for j in range(n):
    rot[j][n-1-i]=mat[i][j]
for row in rot:
  print(*row)
```
