```python
a=['apple','banana','cherry']
b=["orange"]
a.append(b)
print(a)
```

```python
print("copy()")
a=['a','b','c']
a.copy()
print(a)
```

```python
print("count")
a=['a','b','a','d','a','f']
x=a.count('a')
print(x)
```

```python
print("clear")
a=['a','b','c']
a.clear()
print(a)
```

```python
print("extend")
a=['a','b','c']
b=['1','4','8','6']
a.extend(b)
print(a)
```

```python
print("index()")
a=['a','b','c','d']
x=a.index('c')
print(x)
```

```python
print("insert()")
a=['a','k','d','h']
a.insert(2,'g')
print(a)
```

```python
print("pop()")
a=['2','7','5','8']
a.pop(3)
print(a)
```
