```python
s = "a1b10c2"
i = 0
while i < len(s):
    char = s[i]
    i += 1
    num_str = ""
    # Collect all digits following the character
    while i < len(s) and s[i].isdigit():
        num_str += s[i]
        i += 1
    print(char * int(num_str), end="")
```

```python
arr1 = [1, 2, 3, 5]
arr2 = [2, 3, 4, 7]
i = 0
j = 0

while i < len(arr1) and j < len(arr2):
    if arr1[i] < arr2[j]:
        print(arr1[i], end=" ")
        i += 1
    elif arr1[i] > arr2[j]:
        print(arr2[j], end=" ")
        j += 1
    else: # arr1[i] == arr2[j]
        print(arr1[i], end=" ") # Print one if equal
        i += 1
        j += 1
```

```python
n = 12345
sum_prime_digits = 0
while n > 0:
    digit = n % 10
    # Check if digit is 2, 3, 5, or 7
    if digit in [2, 3, 5, 7]:
        sum_prime_digits += digit
    n //= 10
print(f"Sum of prime digits: {sum_prime_digits}")
```

```python
def print_diamond(n):
    # Upper half
    for i in range(n):
        print(" " * (n - i - 1) + "*" * (2 * i + 1))
    # Lower half
    for i in range(n - 2, -1, -1):
        print(" " * (n - i - 1) + "*" * (2 * i + 1))

print_diamond(5)
```

```python
def sort_positioned(arr):
    # evens: index 0, 2, 4... (Ascending)
    evens = sorted(arr[::2])
    # odds: index 1, 3, 5... (Descending)
    odds = sorted(arr[1::2], reverse=True)

    # Merge back into original positions
    arr[::2] = evens
    arr[1::2] = odds
    return arr

print(sort_positioned([13, 2, 4, 15, 12, 10, 5]))
```

```python
def merge_unique(arr1, arr2):
    i, j = 0, 0
    result = []
    while i < len(arr1) and j < len(arr2):
        if arr1[i] < arr2[j]:
            val = arr1[i]
            i += 1
        elif arr1[i] > arr2[j]:
            val = arr2[j]
            j += 1
        else:
            val = arr1[i]
            i += 1
            j += 1

        if not result or result[-1] != val:
            result.append(val)

    # Add remaining elements from both
    result.extend([x for x in arr1[i:] if not result or result[-1] != x])
    result.extend([x for x in arr2[j:] if not result or result[-1] != x])
    return result
```
