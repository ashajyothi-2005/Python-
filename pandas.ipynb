```python
import pandas as pd
data1 = {'Id':[1,2,3],'Name':['rani','teju','asha'],'Age':[30,35,25]}
data2 = {'Id':[4,5,6],'Name':['raju','sanju','bunny'],'Age':[40,45,50]}
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
print(df1)
print(df2)
```

```python
df_combine =pd.concat([df1,df2],ignore_index='true')
print(df_combine)
```

```python
df_filter = df_combine[df_combine['Age'] > 30]
print(df_filter)
```

```python
df_combine['category'] = ['young'if age<40 else 'senior' for age in df_combine['Age']]
print(df_combine)
```
