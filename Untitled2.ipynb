```python
char = input("Enter a character: ")

if 'A' <= char <= 'Z':
    print(f"{char} is uppercase")
elif 'a' <= char <= 'z':
    print(f"{char} is lowercase")
else:
    print(f"{char} is not an alphabet letter")
```

```python
text = input("Enter a message: ")
shift = int(input("Enter shift value: "))

encrypted = ''
for char in text:
    encrypted += chr((ord(char) + shift) % 128)

print("Encrypted message:", encrypted)
```

```python
text = input("Enter encrypted message: ")
shift = int(input("Enter shift value: "))

decrypted = ''
for char in text:
    decrypted += chr((ord(char) - shift) % 128)

print("Decrypted message:", decrypted)
```

```python
rows = int(input("Enter number of rows: "))

for i in range(rows):
    # Print spaces
    for j in range(rows - i - 1):
        print(" ", end=" ")

    # Print characters
    for j in range(i + 1):
        print(chr(65 + j), end=" ")

    print()
```

```python
rows = int(input("Enter number of rows: "))

for i in range(rows):
    for j in range(rows):
        if i == j or j == rows - i - 1:
            print(chr(65 + j), end=' ')
        else:
            print(" ", end=' ')
    print()
```

```python
rows = int(input("Enter number of rows: "))

for i in range(rows):
    num = 1
    print(" " * (rows - i), end='')
    for j in range(i + 1):
        print(num, end=' ')
        num = num * (i - j) // (j + 1)
    print()
```
